```python
import pandas as pd
import pymysql
```

```python
df = pd.read_csv("/workspaces/poetry-demo-test2/clean/no_analysis_result.csv")
df = df.where(pd.notnull(df), None)
df
```

```python
host = '35.194.188.197'
port = 3306
user = 'user'
password = 'password'
db = 'BWA'
charset = 'utf8mb4'

conn = pymysql.connect(host=host, port=port, user=user, passwd=password, db=db, charset=charset)
```

```python
cursor = conn.cursor()
```

```python
sql = """
    update COMMENT
    set
        fake_comment = %s,
        fake_level = %s,
        fake_reason = %s
    where
        comment_id = %s
"""
data = [
    (fake_comment, fake_level, fake_reason, comment_id)
    for comment_id, fake_comment, fake_level, fake_reason in zip(
        df['comment_id'],
        df['fake_comment'],
        df['fake_level'],
        df['fake_reason']
    )
]
cursor.executemany(sql, data)
```
