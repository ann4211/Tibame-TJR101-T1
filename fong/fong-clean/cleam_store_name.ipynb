```python
import pandas as pd
```

```python
df = pd.read_csv("/workspaces/data-pipeline/resource/yizhong_store.csv")
df
```

```python
import re

def clean_store_name(raw_name):
    parts = re.split(r'[-|｜/()（）_\[\]。x]',raw_name)  # list
    store_name = parts[0].strip()  # list取0，字串無空格
    store_name = re.sub(r'[，,]', '', store_name) # 取代，成空格
    return store_name
```

```python
df['names']
```

```python
df['names'].apply(clean_store_name)
```

```python
df['store_name'] = df['names'].apply(clean_store_name)
df
```

```python
def split_address(raw_address):
    address = re.sub(r"^\d{3,6}", '', raw_address) # 取代開頭郵遞區號3-6個數字成空字串
    match = re.match(r"^(.{2,4}[縣市].{1,3}[鄉鎮市區])(.+)", address)  
    # (取代開頭2-4任意字元連接縣市，1-3任意字元連接鄉鎮市區為第一群組) (剩下任意字元為第二組)
    if match:
        area = match.group(1)
        detail = match.group(2)
    else:
        area = ''
        detail = address

    return pd.Series({
        "store_area": area,
        "store_address_clean": detail
        }
    )
```

```python
df['address'].apply(split_address)
```

```python
df[['store_area', 'store_address_clean']] = df['address'].apply(split_address)
df
```
