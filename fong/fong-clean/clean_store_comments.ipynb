```python
import pandas as pd
```

```python
df = pd.read_csv("/workspaces/data-pipeline/resource/20250425_crawler_ruifang_rawdata.csv")
df
```

```python
import re

def clean_store_name(raw_name):
    parts = re.split(r'[-|｜/()（）_\[\]。x]',raw_name)  # list
    store_name = parts[0].strip()  # list取0，字串無空格
    store_name = re.sub(r'[，,]', '', store_name) # 取代，成空格
    return store_name
```

```python
df['店家名稱']
```

```python
df['store_name'] = df['店家名稱'].apply(clean_store_name)
df
```

```python
# 清理評等
df["評分"].str.extract('(\d+)').astype(int)
# 從字串取出數字變形int
```

```python
df['rating_star'] = df["評分"].str.extract('(\d+)').astype(int)
df
```

```python
df['author'] = df['留言者名稱']
df
```

```python
# 清理身分欄位
def prase_user_info(text):
    text = str(text) if not isinstance(text, str) else text
    # 檢查是不是字串，不是的話就是字串類型，才能判斷
    is_local_guide = '在地嚮導' in text
    review_cnt = 0
    photo_cnt = 0
    
    reviews_match = re.search(r"([\d,]+)\s*則評論", text) # 找尋(有多位數字含,為第一組)空格則評論 -> str
    if reviews_match:
        review_cnt = int(reviews_match.group(1).replace(",",""))  # 第一組數字,取代成空字串轉成整數
    else:
        review_cnt == 0

    photos_match = re.search(r"([\d,]+)\s*張相片", text) # 找尋(有多位數字含,為第一組)空格張相片 -> str
    if photos_match:
        photo_cnt = int(photos_match.group(1).replace(",","")) # 第一組數字,取代成空字串轉成整數
    else:
        photo_cnt == 0
    
    return pd.Series({
        "is_local_guide": is_local_guide,
        "review_cnt": review_cnt,
        "photo_cnt": photo_cnt
    })
```

```python
df['留言者身分'].apply(prase_user_info)
```

```python
df[['is_local_guide', 'review_cnt', 'photo_cnt']] = df['留言者身分'].apply(prase_user_info)
df
```

```python
def split_time(text):
    time_match = re.split(r"(\d+)(.+?)前", text)
    time_num = time_match[1]
    time_unit = time_match[2]

    return pd.Series({
        "time_num" : time_num,
        "time_unit" : time_unit
        }
    )
```

```python
df['留言時間'].apply(split_time)
```

```python
df[['time_num', 'time_unit']] = df['留言時間'].apply(split_time)
df
```

```python
df['content_clean'] = df['評論內容']
df
```

```python
df = df.drop(columns=['店家名稱', '評分', '留言者名稱', '留言者身分', '留言時間', '評論內容'])
df
```
