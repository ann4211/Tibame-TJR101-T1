```python
import pandas as pd
```

```python
df = pd.read_csv("/workspaces/poetry-demo-test2/clean/merge_no_no_stid.csv")
```

```python
df = df.drop(columns=['exists_in_b'])
```

```python
df_exist = pd.read_csv("/workspaces/poetry-demo-test2/backup/comment.csv")
```

```python
merge_cols = ["nm_id", "st_id", "user_id", "content_clean"]

# 合併：找出 A 中出現在 B 的列（inner join）
df_common = df.merge(df_exist[merge_cols], on=merge_cols, how='inner')

# 找出 A 中有出現在 B 的列（布林篩選）
df['exists_in_b'] = df[merge_cols].apply(tuple, axis=1).isin(
    df_common[merge_cols].apply(tuple, axis=1)
)
df
```

```python
df = df[df['exists_in_b'] == False]
```

```python
df.to_csv("/workspaces/poetry-demo-test2/clean/merge_no_no_stid.csv", header=True, index=None)
```
