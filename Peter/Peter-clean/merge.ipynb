```python
import pandas as pd
import glob
import os
# 設定 CSV 檔案的路徑
files = glob.glob("C:/Tibame-TJR101-T1/Peter/Data/StoreList/*.csv")  
# 如果需要逐一列出檔案名稱
for file in files:
    print(f"檔案名稱: {file}")
```

```python
# 合併所有 CSV 檔案
try:
    # 讀取並合併所有檔案
    dataframes = [pd.read_csv(file) for file in files]
    merged_df = pd.concat(dataframes, ignore_index=True)
    
    # 儲存合併後的結果為新的 CSV 檔案
    output_path = "C:/Tibame-TJR101-T1/Peter/Data/StoreList/合併後的資料.csv"
    merged_df.to_csv(output_path, index=False)
    print(f"✅ 合併後的資料已成功儲存到: {output_path}")
except Exception as e:
    print(f"⚠️ 合併或儲存檔案時發生錯誤：{e}")
```
